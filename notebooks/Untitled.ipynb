```python
sys_prompt = """
You are a data generator for training a financial analysis reasoning model.

Your job is to create high-quality synthetic training examples that teach a model to:
- Read small financial tables
- Interpret questions about financial performance
- Explain step by step how to reason through a financial dataset to answer complex mathematical and financial questions
- Wrap the question/prompt in <prompt> ... </prompt> tags
- Wrap the paragraphs and tables under <context> ... </context> tags
- Come up with a detailed user intent behind the prompt based on the prompt itself and the paragraphs and tables.
- Perform accurate calculations (margins, variances, ratios, trends)
- Explain reasoning step-by-step
- The user intent, calculations and reasoning should be wrapped under <thinking> ... </thinking> tags
- Produce a clear final answer beginning with: "Final answer:". Be as detailed as possible in explaining the final answer. The whole part with the final answer should be wrapped in <answer> ... </answer> tags

RULES:
- Chain-of-thought must be logically correct.
- End the example with a single “Final answer:” line.
"""
```

```python
user_prompt = """
    Here is the user question: {},
    Here are the paragraphs and tables to serve as your context {},
    Here is the actual evidence that you need to consider from the contexts: {},
    Here is the answer type {},
    Here is the derivation of the answer {},
    Here is the final answer (You need to expound on this):  {}
    Here is the scale at which the answer is expressed: {}
"""
```

```python
prompt = sys_prompt + user_prompt.format(df['questions'][0], df['paragraphs'][0]+"\n"+df['table'][0], df['facts'][0], df['answer_type'][0], df['derivation'][0], df['answer'][0], df['scale'][0])
```
